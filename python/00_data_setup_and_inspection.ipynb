## Q1 Findings

* **Inpatient claims are the largest driver of insurer-paid costs**, with total paid costs of **$1.09 million**.
* Inpatient claims accounted for **99 claims**, compared with 105 outpatient claims and 88 emergency claims.
* Despite not having the highest claim volume, inpatient claims had the **highest average paid amount per claim at $11,034.91**.
* **Emergency claims** were the second-largest contributor to total paid costs, with **$294,441.36** paid across 88 claims.
* **Outpatient, Lab, and Pharmacy claims** contributed substantially less to total paid costs.
* Overall, the analysis suggests that **inpatient care is the primary area of concern because of its high cost per claim**, rather than simply having the highest number of claims.



## Question 2 — CPT & ICD Cost Drivers

**Business question:**
Which medical procedures and diagnoses are driving the insurer's healthcare costs?

**Objective:**
Identify the top 10 CPT (Current Procedural Terminology) codes and top 10 ICD (International Classification of Diseases) codes by total paid amount. Also identify CPT codes with a high average paid amount per claim.

**Business purpose:**
Determine which specific procedures and diagnoses contribute most to healthcare spending and distinguish between cost drivers caused by high claim volume and those caused by high cost per claim.

## Q2 Findings

* The analysis identified the procedures and diagnoses associated with the highest insurer-paid costs.
* **CPT 67890** was the largest individual procedure cost driver, with **$242,735** in total paid costs, followed by CPT 23456 at **$203,790.75**.
* **ICD I10** was the largest diagnosis-level cost driver, with **$259,566** in total paid costs.
* The highest average paid amount per CPT claim was associated with **CPT 36512**, at **$25,600 per claim**.
* Some CPT codes with high average costs per claim did not appear among the top 10 by total paid amount. This indicates that **high cost per claim and high overall spending are not necessarily driven by the same procedures**.
* **CPT 123** appeared in both the top 10 by total paid amount and the top 10 by average paid amount per claim, making it a notable procedure for further investigation.
* Overall, the analysis shows that healthcare spending is concentrated among a relatively small number of procedures and diagnoses, providing potential areas for the insurer to investig
