# Q2 — CPT & ICD Cost Drivers

## Business Question

Which medical procedures and diagnoses are driving insurer healthcare costs?

## Objective

Identify the CPT and ICD codes associated with the highest insurer-paid amounts and determine which CPT codes have the highest average paid amount per claim.

```python
import pandas as pd

```

```python
claims = pd.read_csv('/Users/blessingmashonga/Downloads/healthcare-claims-where-is-the-money-going-tvhlq/claims.csv')
```

## Q2A — Top 10 CPT Codes by Total Paid Amount

CPT (Current Procedural Terminology) codes represent medical procedures and services.

The analysis ranks CPT codes by the total amount paid by the insurer.

```python
top_cpt_paid = (
    claims.groupby('cpt_code')['paid_amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_cpt_paid
```

## Q2B — Top 10 ICD Codes by Total Paid Amount

ICD (International Classification of Diseases) codes represent diagnoses.

The analysis ranks ICD codes by the total amount paid by the insurer.

```python
top_icd_paid = (
    claims.groupby('icd_code')['paid_amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_icd_paid
```

## Q2C — CPT Codes with the Highest Average Paid Amount per Claim

Total paid amount identifies procedures that contribute the most to overall spending.

However, a high total may be caused by a large number of claims.

Average paid amount per claim helps identify procedures that are individually more expensive.

```python
cpt_average_paid = (
    claims.groupby('cpt_code')['paid_amount']
    .mean()
    .sort_values(ascending=False)
    .head(10)
)

cpt_average_paid
```
