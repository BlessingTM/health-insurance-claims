# Q3 — Member-Level Analysis

## Business Question

Which members are driving insurer healthcare spending, and what claim types are responsible for high costs?

## Objective

Identify the highest-cost members based on total insurer-paid amounts and examine the claim types contributing to their healthcare costs.

```python
import pandas as pd
```

```python
claims = pd.read_csv('/Users/blessingmashonga/Downloads/healthcare-claims-where-is-the-money-going-tvhlq/claims.csv')
```

## Q3A — Top 10 Highest-Cost Members

Members are ranked by the total amount paid by the insurer across all of their claims.

```python
top_10_members = (
    claims.groupby('member_id')['paid_amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_10_members
```

## Q3B — Claim Types Driving Costs for the Top 10 Members

The top 10 members are filtered from the claims dataset. Their paid amounts are then broken down by claim type.

```python
top_member_claims = claims[
    claims['member_id'].isin(top_10_members.index)
]

top_member_claims
```

```python
claim_type_by_member = (
    top_member_claims
    .groupby(['member_id', 'claim_type'])['paid_amount']
    .sum()
)

claim_type_by_member
```

## Q3C — Verification

The claim-type totals for each high-cost member should reconcile to the member's overall paid amount.

Member 6 is used as a verification example.

```python
claim_verification = claim_type_by_member.reset_index()

claim_verification[
    claim_verification['member_id'] == 6
]['paid_amount'].sum()
```

```python
## Q3 Summary

This analysis identifies the members contributing the highest total paid amounts and examines the claim types responsible for their costs.

The member-level analysis provides a more detailed view of where healthcare spending is concentrated beyond overall claim-type and procedure-level analysis.
```
