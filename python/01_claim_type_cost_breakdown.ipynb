# Q1 — Claim Type Cost Breakdown

## Business Question

Which claim types are driving the insurer's healthcare costs?

## Objective

Analyze healthcare spending by claim type using total billed amount, total paid amount, claim volume, and average paid amount per claim.

```python
import pandas as pd
```

```python
claims = pd.read_csv('/Users/blessingmashonga/Downloads/healthcare-claims-where-is-the-money-going-tvhlq/claims.csv')
```

```python
claims.shape
```

```python
claims.info()
```

## Q1 Analysis — Cost Breakdown by Claim Type

```python
claim_type_cost_breakdown = claims.groupby('claim_type').agg({
    'billed_amount': 'sum',
    'claim_id': 'count',
    'paid_amount': 'sum'
}).sort_values('paid_amount', ascending=False)

claim_type_cost_breakdown
```

```python
### Average Paid Amount per Claim
```

```python
claims.groupby('claim_type')['paid_amount'].mean()
```
